```python
idade = int(input("Qual é a sua idade: "))
valor_carteira = float(input("Qual é o valor que você tem em carteira? "))

if idade < 16:
  print("Entrada permitida apenas com acompanhante.")
elif idade >= 16 and valor_carteira >= 30:
  print("Ingresso comprado! Bom filme.")
else:
  print("Saldo insuficiente para o ingresso.")
print("Foi bom te atender!!")
```
